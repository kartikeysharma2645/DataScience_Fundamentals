## 1. Data Science Fundamentals

### Data Science Lifecycle

The Data Science lifecycle is a systematic process used to transform a real-world problem into actionable insights or data-driven solutions. It is iterative rather than strictly linear, as findings from later stages may require revisiting earlier stages.

1. **Problem Definition**
   - Clearly identify the business or real-world problem.
   - Define the objectives, scope, constraints, and expected outcomes.

2. **Data Collection**
   - Gather relevant data from sources such as databases, APIs, files, surveys, sensors, or public datasets.
   - Ensure that the collected data is relevant to the problem being investigated.

3. **Data Cleaning and Preparation**
   - Handle missing values, duplicate records, inconsistent formats, and incorrect data.
   - Transform the data into a suitable format for analysis.

4. **Exploratory Data Analysis (EDA)**
   - Examine the structure, distributions, relationships, and patterns within the data.
   - Use statistical measures and visualizations to understand the dataset.

5. **Feature Engineering and Selection**
   - Create useful variables from existing data when necessary.
   - Select relevant features that can improve analysis or predictive performance.

6. **Modeling**
   - Apply appropriate statistical or machine learning techniques based on the problem.
   - Train models using prepared data when prediction or classification is required.

7. **Evaluation**
   - Assess the quality and reliability of the analysis or model using suitable metrics.
   - Compare the results against the original objectives and identify limitations.

8. **Deployment**
   - Put the resulting model, analysis, or solution into practical use when required.
   - Integrate it into an application, workflow, or decision-making process.

9. **Monitoring and Maintenance**
   - Monitor performance after deployment.
   - Update the solution when data, requirements, or real-world conditions change.

### Key Point

The Data Science lifecycle is iterative. A data scientist may return to previous stages when new findings, data-quality issues, changing requirements, or evaluation results indicate that improvements are necessary.

### Types of Data Analysis

Data analysis can be categorized based on the type of question being asked and the objective of the analysis.

1. **Descriptive Analysis**
   - Describes what has happened in the past or what is currently observed in the data.
   - Common techniques include summary statistics, frequency distributions, and data visualization.
   - Example: Calculating the average monthly sales of a company.

2. **Diagnostic Analysis**
   - Investigates why a particular event or outcome occurred.
   - It focuses on identifying relationships, patterns, and potential causes.
   - Example: Analyzing why sales decreased during a particular month.

3. **Predictive Analysis**
   - Uses historical data, statistical methods, and machine learning to estimate future outcomes.
   - Example: Predicting next month's product demand based on historical sales data.

4. **Prescriptive Analysis**
   - Recommends actions that can help achieve a desired outcome.
   - It often uses predictions, optimization techniques, and business constraints.
   - Example: Recommending how much inventory should be ordered to minimize shortages and excess stock.

### Comparison of Analysis Types

| Type | Main Question | Typical Output |
|---|---|---|
| Descriptive | What happened? | Summaries and trends |
| Diagnostic | Why did it happen? | Causes and relationships |
| Predictive | What is likely to happen? | Forecasts and predictions |
| Prescriptive | What should we do? | Recommendations and actions |

## 3. Statistical Foundations

Statistics provides the mathematical foundation for understanding, summarizing, and interpreting data. In Data Science, statistical techniques help identify patterns, measure variability, understand relationships between variables, and make informed conclusions from data.

This section covers fundamental descriptive statistics, probability concepts, percentiles, and correlation.

### Measures of Central Tendency

Measures of central tendency describe the typical or central value of a dataset.

- **Mean:** The arithmetic average of all values. It is calculated by dividing the sum of the values by the number of observations.
- **Median:** The middle value when the observations are arranged in ascending or descending order. If there is an even number of observations, the median is the average of the two middle observations.
- **Mode:** The value that occurs most frequently in a dataset.

The mean can be strongly affected by extreme values, whereas the median is generally more robust to outliers.

```python
sales = (12500, 14800, 13200, 15600, 14100, 16700, 13900)
mean = sum(sales)/ len(sales)
print(f"Mean: {mean}")
```

### Median

The median is the middle value of a dataset after the observations have been arranged in ascending or descending order.

- If the dataset contains an odd number of observations, the median is the single middle value.
- If the dataset contains an even number of observations, the median is the average of the two middle values.

The median is less sensitive to extreme values than the mean, making it useful for datasets that may contain outliers.

```python
sales = (12500, 14800, 13200, 15600, 14100, 16700, 13900)  
middle_index = len(sales)//2
sorted_sales = sorted(sales)
median = sorted_sales[middle_index]
print(f"Median: {median}")
```

### Mode

The mode is the value that occurs most frequently in a dataset. A dataset can have one mode, multiple modes, or no mode when no value occurs more frequently than the others.

The mode is particularly useful when identifying the most common category or value in a dataset.

```python
values = [2, 3, 4, 3, 5, 3, 6, 4, 2, 3]
sorted_values = sorted(values)
frequency = {}
for i in sorted_values:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1
print(frequency)
for j in frequency:
    if frequency[j] == max(frequency.values()):
        mode = j
print(f"Mode: {mode}")
```

### Interpretation

For the weekly sales dataset, the mean sales value represents the average daily sales, while the median represents the middle daily sales value after sorting the observations.

The mode is useful when a dataset contains repeated values. Since each sales value in this dataset occurs only once, there is no meaningful mode.

Comparing the mean and median can also provide an initial indication of the distribution of the data. When the two values are relatively close, the dataset may have limited skewness, although additional statistical analysis is required to make a reliable conclusion.

### Measures of Dispersion

Measures of dispersion describe how widely the observations are spread around a central value. While measures of central tendency describe the center of a dataset, measures of dispersion help understand the variability within the data.

The major measures of dispersion covered in this section are:

- **Range:** The difference between the maximum and minimum values in a dataset.
- **Variance:** Measures the average squared deviation of observations from the mean.
- **Standard Deviation:** The square root of variance and represents dispersion in the same units as the original data.

Understanding dispersion is important because two datasets can have the same mean but significantly different levels of variability.

### Range

The range is the simplest measure of dispersion. It represents the difference between the largest and smallest observations in a dataset.

\[
\text{Range} = \text{Maximum Value} - \text{Minimum Value}
\]

A larger range indicates that the observations are more widely spread, while a smaller range indicates that they are closer together.

For the weekly sales dataset, the range is calculated using the maximum and minimum daily sales values.

```python
sales = (12500, 14800, 13200, 15600, 14100, 16700, 13900)
range = max(sales) - min(sales)
print(f"Range: {range}")
```

### Variance

Variance measures how much the observations in a dataset differ from their mean. It is calculated by finding the deviation of each observation from the mean, squaring each deviation, and then calculating the average of the squared deviations.

For a population dataset, variance is calculated using:

\[
\sigma^2 = \frac{\sum (x_i-\mu)^2}{N}
\]

where:

- \(x_i\) represents each observation.
- \(\mu\) represents the population mean.
- \(N\) represents the total number of observations.

A larger variance indicates greater variability in the dataset, while a smaller variance indicates that the observations are closer to the mean.

For this exercise, the variance is calculated manually using Python to demonstrate the underlying statistical process.

```python
sales = (12500, 14800, 13200, 15600, 14100, 16700, 13900)
mean = sum(sales)/ len(sales)
squared_deviations = []
for value in sales:
    deviation = value - mean
    squared_deviation = deviation ** 2
    squared_deviations.append(squared_deviation)

variance = sum(squared_deviations) / len(sales)
print(f"Variance: {variance:.2f}")
```

### Standard Deviation

Standard deviation measures the amount of variation or dispersion in a dataset. It indicates how far the observations typically lie from the mean.

Standard deviation is calculated as the square root of variance:

\[
\text{Standard Deviation} = \sqrt{\text{Variance}}
\]

Unlike variance, which is expressed in squared units, standard deviation is expressed in the **same units as the original data**, making it easier to interpret.

A smaller standard deviation indicates that the observations are generally closer to the mean, while a larger standard deviation indicates greater variability.

For the weekly sales dataset, standard deviation helps quantify how much the daily sales values typically vary around the average daily sales.

```python
import math
std_deviation = math.sqrt(variance)
print(f"Standard Deviation: {std_deviation:.2f}")
```

### Probability Basics

Probability measures the likelihood of an event occurring. Its value ranges from 0 to 1, where 0 represents an impossible event and 1 represents a certain event.

The probability of an event is calculated as:

\[
P(A) = \frac{\text{Number of favorable outcomes}}{\text{Total number of possible outcomes}}
\]

For example, when a fair six-sided die is rolled, there are six possible outcomes: 1, 2, 3, 4, 5, and 6. The probability of rolling an even number is determined by the favorable outcomes \(2, 4,\) and \(6\).

Probability concepts are useful in Data Science for uncertainty estimation, risk analysis, statistical modeling, and decision-making.

```python
dice = {1, 2, 3, 4, 5, 6}
even_number = {2, 4, 6}
probability = len(even_number)/len(dice)
print(f"Probability of rolling an even number: {probability:.2f}")
```

### Conditional Probability

Conditional probability measures the probability of an event occurring given that another event has already occurred.

It is represented as:

\[
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
\]

where:

- \(P(A \mid B)\) is the probability of event \(A\) occurring given event \(B\).
- \(P(A \cap B)\) is the probability that both events \(A\) and \(B\) occur.
- \(P(B)\) is the probability of event \(B\).

Conditional probability is widely used in Data Science applications such as classification, risk assessment, medical diagnosis, and Bayesian analysis.

For example, when drawing a card from a standard deck, the probability of drawing a King given that the selected card is a face card can be calculated using conditional probability.

```python
face_cards = {"Jack": 4, "Queen": 4, "King": 4}
kings_face_cards = face_cards["King"]

conditional_probability = kings_face_cards / sum(face_cards.values())

print(f"Conditional Probability of drawing a King given that a face card is drawn: {conditional_probability:.2f}")
```

### Percentiles

Percentiles describe the relative position of a value within a dataset. The \(p\)-th percentile is the value below which approximately \(p\%\) of the observations fall.

For example:

- The **25th percentile (Q1)** represents the point below which approximately 25% of the observations fall.
- The **50th percentile (Q2)** corresponds to the median.
- The **75th percentile (Q3)** represents the point below which approximately 75% of the observations fall.

Percentiles are useful for understanding the distribution of data, identifying unusually high or low observations, and comparing an individual observation with the rest of a dataset.

In Data Science, percentiles are commonly used in exploratory data analysis, performance analysis, income analysis, and outlier detection.

```python
import statistics

sales = (12500, 14800, 13200, 15600, 14100, 16700, 13900)

median_sales = statistics.median(sales)

print(f"50th percentile (median): {median_sales}")
```

```python
import numpy as np
sales = (12500, 14800, 13200, 15600, 14100, 16700, 13900)
percentile_75 = np.percentile(sales, 75)

print(f"75th percentile (Q3): {percentile_75}")
```

### Correlation

Correlation measures the strength and direction of the relationship between two numerical variables.

- **Positive correlation:** Both variables tend to increase together.
- **Negative correlation:** One variable tends to increase as the other decreases.
- **Zero or weak correlation:** Little or no linear relationship.

The Pearson correlation coefficient ranges from -1 to +1.

```python
hours_studied = [2, 3, 4, 5, 6, 7]
exam_scores = [50, 55, 65, 70, 78, 85]
correlation_matrix = np.corrcoef(hours_studied, exam_scores)

print(f"Correlation Matrix:\n{correlation_matrix}")
```

```python
correlation_coefficient = correlation_matrix[0, 1]
print(f"Pearson's correlation coefficient: {correlation_coefficient:.2f}")
```

## 4. Data Types and Structures

Understanding data types is essential for selecting appropriate analysis and modeling techniques.

### Numerical Data

Numerical data represents measurable quantities and can be divided into:

- **Discrete:** Countable values, such as number of customers or products sold.
- **Continuous:** Measurable values that can take a range of values, such as height, weight, or temperature.

### Categorical Data

Categorical data represents groups or labels rather than numerical measurements.

- **Nominal:** Categories without a natural order, such as gender, city, or product type.
- **Ordinal:** Categories with a meaningful order, such as satisfaction levels or education levels.

```python
data_types = {
    "age": "Discrete",
    "height": "Continuous",
    "number_of_courses": "Discrete",
    "city": "Nominal",
    "satisfaction": "Ordinal"
}

for variable, classification in data_types.items():
    print(f"{variable}: {classification}")
```
