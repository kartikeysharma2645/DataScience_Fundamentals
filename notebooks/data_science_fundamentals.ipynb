## 1. Data Science Fundamentals

### Data Science Lifecycle

The Data Science lifecycle is a systematic process used to transform a real-world problem into actionable insights or data-driven solutions. It is iterative rather than strictly linear, as findings from later stages may require revisiting earlier stages.

1. **Problem Definition**
   - Clearly identify the business or real-world problem.
   - Define the objectives, scope, constraints, and expected outcomes.

2. **Data Collection**
   - Gather relevant data from sources such as databases, APIs, files, surveys, sensors, or public datasets.
   - Ensure that the collected data is relevant to the problem being investigated.

3. **Data Cleaning and Preparation**
   - Handle missing values, duplicate records, inconsistent formats, and incorrect data.
   - Transform the data into a suitable format for analysis.

4. **Exploratory Data Analysis (EDA)**
   - Examine the structure, distributions, relationships, and patterns within the data.
   - Use statistical measures and visualizations to understand the dataset.

5. **Feature Engineering and Selection**
   - Create useful variables from existing data when necessary.
   - Select relevant features that can improve analysis or predictive performance.

6. **Modeling**
   - Apply appropriate statistical or machine learning techniques based on the problem.
   - Train models using prepared data when prediction or classification is required.

7. **Evaluation**
   - Assess the quality and reliability of the analysis or model using suitable metrics.
   - Compare the results against the original objectives and identify limitations.

8. **Deployment**
   - Put the resulting model, analysis, or solution into practical use when required.
   - Integrate it into an application, workflow, or decision-making process.

9. **Monitoring and Maintenance**
   - Monitor performance after deployment.
   - Update the solution when data, requirements, or real-world conditions change.

### Key Point

The Data Science lifecycle is iterative. A data scientist may return to previous stages when new findings, data-quality issues, changing requirements, or evaluation results indicate that improvements are necessary.

### Types of Data Analysis

Data analysis can be categorized based on the type of question being asked and the objective of the analysis.

1. **Descriptive Analysis**
   - Describes what has happened in the past or what is currently observed in the data.
   - Common techniques include summary statistics, frequency distributions, and data visualization.
   - Example: Calculating the average monthly sales of a company.

2. **Diagnostic Analysis**
   - Investigates why a particular event or outcome occurred.
   - It focuses on identifying relationships, patterns, and potential causes.
   - Example: Analyzing why sales decreased during a particular month.

3. **Predictive Analysis**
   - Uses historical data, statistical methods, and machine learning to estimate future outcomes.
   - Example: Predicting next month's product demand based on historical sales data.

4. **Prescriptive Analysis**
   - Recommends actions that can help achieve a desired outcome.
   - It often uses predictions, optimization techniques, and business constraints.
   - Example: Recommending how much inventory should be ordered to minimize shortages and excess stock.

### Comparison of Analysis Types

| Type | Main Question | Typical Output |
|---|---|---|
| Descriptive | What happened? | Summaries and trends |
| Diagnostic | Why did it happen? | Causes and relationships |
| Predictive | What is likely to happen? | Forecasts and predictions |
| Prescriptive | What should we do? | Recommendations and actions |

### Role of Data Science in Solving Real-World Problems

Data Science helps organizations convert raw data into useful information and actionable insights. A typical solution involves collecting relevant data, preparing and analyzing it, identifying meaningful patterns, and using the results to support decisions or build predictive systems.

The value of Data Science is not simply in producing predictions or visualizations. The analysis must address a clearly defined problem, use appropriate and reliable data, and produce results that can be interpreted in the context of the real-world situation.

### Python Operators

Python provides operators for performing arithmetic calculations, comparisons, logical operations, and assignments. These operators are fundamental for manipulating and evaluating data.

```python
a = 6
b = 9
print(a + b)
print(a - b)
print(a > b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
print(a < b)
print(a > b)
print(a <= b)
print(a >= b)
print(a == b)
print(a > 5 and b < 20)
print(a > 5 or b < 20)
print(a != b)
print(not (a > b))
```

### Conditional Statements

Conditional statements allow a Python program to make decisions based on whether specified conditions are true or false. The primary constructs are `if`, `elif`, and `else`.

```python
marks = int(input("Enter your marks: "))
if marks < 0 or marks > 100: 
    print("Invalid marks, please enter marks between 0 to 100")
elif marks >= 90 and marks <= 100:
    print("Excellent")
elif marks >= 75 and marks < 90:
    print("Very Good")
elif marks >= 60 and marks < 75:
    print("Good")
elif marks >= 40:
    print("Pass")
elif marks >=0 and marks < 40: 
    print("Fail")
```
